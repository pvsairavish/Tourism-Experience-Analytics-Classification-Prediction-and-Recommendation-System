## Load Data

````python
# ====================== LOAD DATASET ======================

DATA_PATH = "/kaggle/input/datasets/sairavishpv/tourism-dataset/"

transaction = pd.read_excel(DATA_PATH + "Transaction.xlsx")
user = pd.read_excel(DATA_PATH + "User.xlsx")
item = pd.read_excel(DATA_PATH + "Item.xlsx")
city = pd.read_excel(DATA_PATH + "City.xlsx")
country = pd.read_excel(DATA_PATH + "Country.xlsx")
region = pd.read_excel(DATA_PATH + "Region.xlsx")
continent = pd.read_excel(DATA_PATH + "Continent.xlsx")
attraction_type = pd.read_excel(DATA_PATH + "Type.xlsx")
visit_mode = pd.read_excel(DATA_PATH + "Mode.xlsx")

print("✅ All files loaded successfully!")
print("Transaction:", transaction.shape)
print("User:", user.shape)
print("Item:", item.shape)
````

## Full Data Cleaning & Merging

````python
# ====================== DATA CLEANING & MERGING ======================
print("Starting data cleaning and merging...")

# Clean Transaction
transaction = transaction.dropna(subset=['Rating', 'UserId', 'AttractionId'])
transaction['VisitYear'] = transaction['VisitYear'].astype(int)
transaction['VisitMonth'] = transaction['VisitMonth'].astype(int)
transaction['Rating'] = pd.to_numeric(transaction['Rating'], errors='coerce')
transaction = transaction.dropna(subset=['Rating'])

# Merge step-by-step
df = transaction.merge(user, on='UserId', how='left')
df = df.merge(continent, on='ContinentId', how='left')
df = df.merge(region, on='RegionId', how='left')
df = df.merge(country, on='CountryId', how='left')
df = df.merge(city, on='CityId', how='left')
df = df.merge(item, on='AttractionId', how='left')
df = df.merge(attraction_type, on='AttractionTypeId', how='left')

# Clean VisitMode
if df['VisitMode'].dtype in [np.int64, np.float64]:
    mode_map = dict(zip(visit_mode['VisitModeId'], visit_mode['VisitMode']))
    df['VisitMode'] = df['VisitMode'].map(mode_map)
df['VisitMode'] = df['VisitMode'].astype(str).str.strip()

# Final cleaning
df = df.dropna(subset=['Continent', 'Country', 'AttractionType', 'VisitMode', 'Attraction'])
df = df.drop_duplicates()

print("\n✅ Final Master Dataset Shape:", df.shape)
print("Missing values remaining:", df.isnull().sum().sum())
df.head()
````

# **Feature Engineering**

````python
# ====================== IMPROVED FEATURE ENGINEERING ======================
model_df = df.copy()

# ---------- Label Encoding ----------
le_dict = {}
cat_cols = ['VisitMode', 'Continent', 'Region', 'Country', 'CityName', 'AttractionType']

for col in cat_cols:
    le = LabelEncoder()
    model_df[col + '_encoded'] = le.fit_transform(model_df[col].astype(str))
    le_dict[col] = le

# ---------- Strong User-level Features ----------
user_agg = model_df.groupby('UserId').agg({
    'Rating': ['mean', 'count', 'std', 'min', 'max'],
    'VisitMode_encoded': ['nunique', 'mean'],
    'AttractionType_encoded': 'nunique',
    'VisitMonth': ['mean', 'nunique']
}).reset_index()

user_agg.columns = [
    'UserId',
    'UserAvgRating', 'UserVisitCount', 'UserRatingStd', 'UserMinRating', 'UserMaxRating',
    'UserModeDiversity', 'UserModeMean',
    'UserTypeDiversity', 'UserAvgMonth', 'UserMonthDiversity'
]
user_agg['UserRatingStd'] = user_agg['UserRatingStd'].fillna(0)

# ---------- Strong Attraction-level Features ----------
attr_agg = model_df.groupby('AttractionId').agg({
    'Rating': ['mean', 'count', 'std'],
    'VisitMode_encoded': 'nunique',
    'UserId': 'nunique'
}).reset_index()

attr_agg.columns = [
    'AttractionId',
    'AttractionAvgRating', 'AttractionPopularity', 'AttractionRatingStd',
    'AttractionModeDiversity', 'AttractionUniqueUsers'
]
attr_agg['AttractionRatingStd'] = attr_agg['AttractionRatingStd'].fillna(0)

# Merge features
model_df = model_df.merge(user_agg, on='UserId', how='left')
model_df = model_df.merge(attr_agg, on='AttractionId', how='left')

# ---------- Extra Engineered Features ----------
model_df['IsPeakSeason'] = model_df['VisitMonth'].isin([5, 6, 7, 8, 12]).astype(int)
model_df['Rating_User_Attr_Diff'] = model_df['UserAvgRating'] - model_df['AttractionAvgRating']
model_df['UserLoyalty'] = model_df['UserVisitCount'] / (model_df['UserVisitCount'].max() + 1)
model_df['AttractionPopularityLog'] = np.log1p(model_df['AttractionPopularity'])

# ---------- Final Feature List ----------
feature_cols = [
    'VisitYear', 'VisitMonth', 'IsPeakSeason',
    'Continent_encoded', 'Region_encoded', 'Country_encoded', 'CityName_encoded',
    'AttractionType_encoded',
    'UserAvgRating', 'UserVisitCount', 'UserRatingStd', 'UserMinRating', 'UserMaxRating',
    'UserModeDiversity', 'UserModeMean', 'UserTypeDiversity', 'UserAvgMonth', 'UserMonthDiversity',
    'AttractionAvgRating', 'AttractionPopularity', 'AttractionRatingStd',
    'AttractionModeDiversity', 'AttractionUniqueUsers', 'AttractionPopularityLog',
    'Rating_User_Attr_Diff', 'UserLoyalty'
]

X = model_df[feature_cols]
y_clf = model_df['VisitMode_encoded']
y_reg = model_df['Rating']

print("✅ Feature Engineering Completed")
print("Total Features:", len(feature_cols))
print("Shape:", X.shape)
````

## Train-Test Split

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_clf_train, y_clf_test, y_reg_train, y_reg_test = train_test_split(
    X, y_clf, y_reg, test_size=0.20, random_state=42, stratify=y_clf
)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)
````

## Classification Models (Predict Visit Mode)

````python
# ====================== CLASSIFICATION + MLFLOW ======================
import mlflow
import mlflow.sklearn
import mlflow.xgboost
import mlflow.lightgbm

mlflow.set_experiment("Tourism_Experience_Analytics")

classification_models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=300, max_depth=16, min_samples_split=4,
        class_weight='balanced', random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBClassifier(
        n_estimators=300, max_depth=9, learning_rate=0.07,
        subsample=0.85, colsample_bytree=0.85,
        random_state=42, eval_metric='mlogloss', n_jobs=-1
    ),
    "LightGBM": LGBMClassifier(
        n_estimators=300, max_depth=9, learning_rate=0.07,
        subsample=0.85, colsample_bytree=0.85,
        class_weight='balanced', random_state=42, n_jobs=-1
    )
}

class_results = {}

for model_name, model in classification_models.items():
    with mlflow.start_run(run_name=f"Classification_{model_name}"):
        
        model.fit(X_train, y_clf_train)
        y_pred = model.predict(X_test)
        
        acc = accuracy_score(y_clf_test, y_pred)
        f1 = f1_score(y_clf_test, y_pred, average='weighted')
        precision = precision_score(y_clf_test, y_pred, average='weighted')
        recall = recall_score(y_clf_test, y_pred, average='weighted')
        
        mlflow.log_param("model_type", model_name)
        mlflow.log_param("problem_type", "Classification")
        mlflow.log_metric("Accuracy", acc)
        mlflow.log_metric("F1_Weighted", f1)
        mlflow.log_metric("Precision", precision)
        mlflow.log_metric("Recall", recall)
        
        if model_name == "XGBoost":
            mlflow.xgboost.log_model(model, "model")
        elif model_name == "LightGBM":
            mlflow.lightgbm.log_model(model, "model")
        else:
            mlflow.sklearn.log_model(model, "model")
        
        class_results[model_name] = {
            "Accuracy": round(acc, 4),
            "F1_Weighted": round(f1, 4),
            "Precision": round(precision, 4),
            "Recall": round(recall, 4)
        }
        print(f"{model_name:12} → Accuracy: {acc:.4f} | F1: {f1:.4f}")

print("\n===== Classification Results =====")
display(pd.DataFrame(class_results).T.sort_values("Accuracy", ascending=False))

best_clf_name = pd.DataFrame(class_results).T['Accuracy'].idxmax()
best_clf_model = classification_models[best_clf_name]
print(f"\n🏆 Best Classification Model: {best_clf_name}")
````

## Regression Models (Predict Rating)

````python
# ====================== REGRESSION + MLFLOW ======================
regression_models = {
    "RandomForest": RandomForestRegressor(
        n_estimators=300, max_depth=16, min_samples_split=4,
        random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300, max_depth=8, learning_rate=0.07,
        subsample=0.85, colsample_bytree=0.85,
        random_state=42, n_jobs=-1
    ),
    "LightGBM": LGBMRegressor(
        n_estimators=300, max_depth=8, learning_rate=0.07,
        subsample=0.85, colsample_bytree=0.85,
        random_state=42, n_jobs=-1
    )
}

reg_results = {}

for model_name, model in regression_models.items():
    with mlflow.start_run(run_name=f"Regression_{model_name}"):
        
        model.fit(X_train, y_reg_train)
        y_pred = model.predict(X_test)
        
        rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred))
        mae = mean_absolute_error(y_reg_test, y_pred)
        r2 = r2_score(y_reg_test, y_pred)
        
        mlflow.log_param("model_type", model_name)
        mlflow.log_param("problem_type", "Regression")
        mlflow.log_metric("RMSE", rmse)
        mlflow.log_metric("MAE", mae)
        mlflow.log_metric("R2_Score", r2)
        
        if model_name == "XGBoost":
            mlflow.xgboost.log_model(model, "model")
        elif model_name == "LightGBM":
            mlflow.lightgbm.log_model(model, "model")
        else:
            mlflow.sklearn.log_model(model, "model")
        
        reg_results[model_name] = {
            "RMSE": round(rmse, 4),
            "MAE": round(mae, 4),
            "R2": round(r2, 4)
        }
        print(f"{model_name:12} → RMSE: {rmse:.4f} | MAE: {mae:.4f} | R²: {r2:.4f}")

print("\n===== Regression Results =====")
display(pd.DataFrame(reg_results).T.sort_values("R2", ascending=False))

best_reg_name = pd.DataFrame(reg_results).T['R2'].idxmax()
best_reg_model = regression_models[best_reg_name]
print(f"\n🏆 Best Regression Model: {best_reg_name}")
````

## Recommendation System

````python
# ====================== RECOMMENDATION SYSTEM ======================
from surprise import Dataset, Reader, SVD
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Collaborative Filtering (SVD)
reader = Reader(rating_scale=(1, 5))
data_surprise = Dataset.load_from_df(df[['UserId', 'AttractionId', 'Rating']], reader)

svd = SVD(n_factors=80, n_epochs=25, lr_all=0.005, reg_all=0.02, random_state=42)
trainset = data_surprise.build_full_trainset()
svd.fit(trainset)
print("✅ Collaborative Filtering (SVD) trained!")

# 2. Content-Based Filtering
attraction_features = df.groupby('AttractionId').agg({
    'Attraction': 'first',
    'AttractionType': 'first',
    'CityName': 'first'
}).reset_index()

attraction_features['content'] = (
    attraction_features['AttractionType'].astype(str) + " " +
    attraction_features['CityName'].astype(str)
)

tfidf = TfidfVectorizer(stop_words='english', max_features=500)
tfidf_matrix = tfidf.fit_transform(attraction_features['content'])
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

def get_content_recommendations(attraction_id, top_n=6):
    try:
        idx = attraction_features[attraction_features['AttractionId'] == attraction_id].index[0]
        sim_scores = sorted(list(enumerate(cosine_sim[idx])), key=lambda x: x[1], reverse=True)[1:top_n+1]
        indices = [i[0] for i in sim_scores]
        return attraction_features.iloc[indices][['AttractionId', 'Attraction', 'AttractionType', 'CityName']]
    except:
        return pd.DataFrame()

print("✅ Content-Based Filtering ready!")
print("\nSample Recommendations:")
display(get_content_recommendations(df['AttractionId'].iloc[0]))
````
